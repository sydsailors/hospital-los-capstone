```python
# Load and verify raw datasets
import pandas as pd
import numpy as np
from pathlib import Path

inpatient = pd.read_excel("../data/raw/h254d.xlsx")
patients = pd.read_excel("../data/raw/h256.xlsx")

print("HC-254D shape:", inpatient.shape)
print("HC-256 shape:", patients.shape)
```

```python
# Select the columns from each dataset needed for analysis
inpatient_columns = [
    "DUPERSID",
    "NUMNIGHX",
    "EMERROOM",
    "RSNINHOS",
    "ANYOPER",
    "IPXP24X",
    "IPTC24X"
]

patient_columns = [
    "DUPERSID",
    "AGE24X",
    "SEX",
    "RACETHX",
    "RTHLTH53",
    "MNHLTH53",
    "INSCOV24",
    "REGION24",
    "PERWT24F",
    "DIABDX_M18",
    "HIBPDX",
    "CHDDX",
    "STRKDX",
    "ASTHDX"
]

inpatient_selected = inpatient[inpatient_columns].copy()
patients_selected = patients[patient_columns].copy()

print("Selected inpatient data:", inpatient_selected.shape)
print("Selected patient data:", patients_selected.shape)
```

```python
# Inspect data types
print("Inpatient data types:")
print(inpatient_selected.dtypes)

print("\nPatient data types:")
print(patients_selected.dtypes)
```

```python
# Inspect missing values 
print("Inpatient missing values:")
print(inpatient_selected.isna().sum())

print("\nPatient missing values:")
print(patients_selected.isna().sum())
```

```python
# Inspect categorical variables
categorical_columns = [
    "EMERROOM",
    "RSNINHOS",
    "ANYOPER",
    "SEX",
    "RACETHX",
    "RTHLTH53",
    "MNHLTH53",
    "INSCOV24",
    "REGION24",
    "DIABDX_M18",
    "HIBPDX",
    "CHDDX",
    "STRKDX",
    "ASTHDX"
]

for col in categorical_columns:
    if col in inpatient_selected.columns:
        print(f"\n--- {col} ---")
        print(inpatient_selected[col].value_counts().sort_index())

    if col in patients_selected.columns:
        print(f"\n--- {col} ---")
        print(patients_selected[col].value_counts().sort_index())
```

```python
# Inspect minimums and maximums of numeric values
numeric_columns = [
    "NUMNIGHX",
    "IPXP24X",
    "IPTC24X"
]

print(inpatient_selected[numeric_columns].agg(["min", "max"]))
```

```python
# Inspect minimums and maximums of age and weight
print(patients_selected[
    ["AGE24X", "PERWT24F"]
].agg(["min", "max"]))
```

```python
# Show only negative values in selected inpatient variables
print("NEGATIVE CODES - INPATIENT")

for col in inpatient_selected.columns:
    if pd.api.types.is_numeric_dtype(inpatient_selected[col]):
        negative = inpatient_selected.loc[
            inpatient_selected[col] < 0, col
        ].value_counts().sort_index()

        if not negative.empty:
            print(f"\n{col}:")
            print(negative)


print("\nNEGATIVE CODES - PATIENT")

for col in patients_selected.columns:
    if pd.api.types.is_numeric_dtype(patients_selected[col]):
        negative = patients_selected.loc[
            patients_selected[col] < 0, col
        ].value_counts().sort_index()

        if not negative.empty:
            print(f"\n{col}:")
            print(negative)
```

```python
def summarize_special_codes(df, dataset_name):
    results = []

    for col in df.columns:
        if col == "DUPERSID":
            continue

        if pd.api.types.is_numeric_dtype(df[col]):
            special = df.loc[df[col] < 0, col].value_counts()

            for code, count in special.items():
                results.append({
                    "Dataset": dataset_name,
                    "Attribute": col,
                    "Special Code": int(code),
                    "Count": int(count)
                })

    return pd.DataFrame(
        results,
        columns=["Dataset", "Attribute", "Special Code", "Count"]
    )

special_codes = pd.concat([
    summarize_special_codes(inpatient_selected, "HC-254D"),
    summarize_special_codes(patients_selected, "HC-256")
], ignore_index=True)

special_codes = special_codes.sort_values(
    ["Dataset", "Attribute", "Special Code"]
)

print(special_codes.to_string(index=False))
```

```python
# Preserve original selected data
inpatient_clean = inpatient_selected.copy()
patients_clean = patients_selected.copy()

# MEPS special response codes
special_codes = [-1, -7, -8, -15]

# Replace special codes only in columns where they occur
inpatient_clean[["RSNINHOS", "ANYOPER"]] = (
    inpatient_clean[["RSNINHOS", "ANYOPER"]]
    .replace(special_codes, np.nan)
)

patient_special_columns = [
    "AGE24X",
    "ASTHDX",
    "CHDDX",
    "DIABDX_M18",
    "HIBPDX",
    "MNHLTH53",
    "REGION24",
    "RTHLTH53",
    "STRKDX"
]

patients_clean[patient_special_columns] = (
    patients_clean[patient_special_columns]
    .replace(special_codes, np.nan)
)

print("Inpatient shape:", inpatient_clean.shape)
print("Patient shape:", patients_clean.shape)
```

```python
print("Inpatient missing values:")
print(inpatient_clean.isna().sum())

print("\nPatient missing values:")
print(patients_clean.isna().sum())
```

```python
print(
    "Duplicate patient IDs:",
    patients_clean["DUPERSID"].duplicated().sum()
)

print(
    "Duplicate inpatient person IDs:",
    inpatient_clean["DUPERSID"].duplicated().sum()
)

print(
    "Missing inpatient LOS:",
    inpatient_clean["NUMNIGHX"].isna().sum()
)
```

```python
# Merge the datasets with DUPERSID
merged = inpatient_clean.merge(
    patients_clean,
    on="DUPERSID",
    how="left",
    validate="many_to_one",
    indicator=True
)

print("Merged dataset shape:", merged.shape)

print("\nMerge results:")
print(merged["_merge"].value_counts())
```

```python
print("Missing values after merging:")
print(merged.isna().sum().sort_values(ascending=False))
```

```python
# Final data validation

print("Final dataset shape:", merged_clean.shape)

print("\nMissing dependent variable:")
print(merged_clean["NUMNIGHX"].isna().sum())

print("\nDuplicate complete rows:")
print(merged_clean.duplicated().sum())

print("\nRemaining negative special codes:")
for col in merged_clean.select_dtypes(include="number").columns:
    if col != "DUPERSID":
        count = (merged_clean[col] < 0).sum()
        if count > 0:
            print(col, count)

print("\nMissing values by attribute:")
print(merged_clean.isna().sum().sort_values(ascending=False))
```

```python
# Remove temporary merge indicator
merged_clean = merged.drop(columns=["_merge"]).copy()

# Create output folder
output_dir = Path("../data/processed")
output_dir.mkdir(parents=True, exist_ok=True)

# Save cleaned and merged dataset
output_file = output_dir / "meps_los_clean.csv"
merged_clean.to_csv(output_file, index=False)

print("Final dataset shape:", merged_clean.shape)
print("Saved to:", output_file)
```
